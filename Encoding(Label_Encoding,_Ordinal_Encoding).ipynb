```python
# Encoding --> transform your categorical data into numnerical data
```

```python
import pandas as pd
```

```python
df = pd.read_csv('/content/covid_toy.csv')
```

```python
df.head(2)
```

```python
# 1. Label Encoding --> it is apply only single column.
```

```python
df['fever'] = df['fever'].fillna(df['fever'].mean())
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
df['gender']=lb.fit_transform(df['gender'])
df['cough']=lb.fit_transform(df['cough'])
df['city']=lb.fit_transform(df['city'])
df['has_covid']=lb.fit_transform(df['has_covid'])
```

```python
df.head(2)
```

```python
x=df.drop(columns=['has_covid'])
y=df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import numpy as np
```

```python
np.round(x_train.describe(),2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
x_train_sc=sc.fit_transform(x_train)
```

```python
x_train_new=pd.DataFrame(x_train_sc,columns=x_train.columns)
```

```python
np.round(x_train_new.describe(),2)
```

```python
# 2. Ordinal Encoder --> we can set the value of sub-categories.
# df['gender'] --> we can set Male=0 , Female=1
```

```python
df=pd.read_csv("/content/covid_toy.csv")
```

```python
df.head(2)
```

```python
df['fever']=df['fever'].fillna(df['fever'].mean())
```

```python
df=df[['gender','cough','city','has_covid']]
```

```python
df.head(2)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe= OrdinalEncoder(categories=[
    ['Male','Female'],
    ['Mild','Strong'],
    ['Kolkata','Bangalore','Delhi','Mumbai'],
    ['No','Yes']
])
```

```python
df_arr = oe.fit_transform(df)
df_arr
```

```python
df_new=pd.DataFrame(df_arr,columns = df.columns)
df_new
```

```python
df_new=df_new.astype(int)
df_new
```

```python
X=df_new.drop(columns=['has_covid'])
Y=df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test= train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
import numpy as np
```

```python
np.round(X_train.describe(),2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
X_train_sc=sc.fit_transform(X_train)
```

```python
X_train_new=pd.DataFrame(X_train_sc,columns=X_train.columns)
```

```python
np.round(X_train_new.describe(),2)
```
