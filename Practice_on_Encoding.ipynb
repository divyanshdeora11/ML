```python
import numpy as np
import pandas as pd
```

```python
df_original = pd.read_csv('/content/Churn_Modelling.csv')
```

```python
df_original.head()
```

```python
df_original.isnull().sum()
```

```python
# 1. Label Encoding
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb = LabelEncoder()
```

```python
df_original['Surname'] = lb.fit_transform(df_original['Surname'])
df_original['Geography'] = lb.fit_transform(df_original['Geography'])
df_original['Gender'] = lb.fit_transform(df_original['Gender'])
df_original['CustomerId'] = lb.fit_transform(df_original['CustomerId'])
```

```python
df_original.head()
```

```python
# stand
```

```python
x = df_original.drop(columns=['Exited'])
y = df_original['Exited']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
x_train_sc = sc.fit_transform(x_train)
x_test_sc = sc.transform(x_test)
```

```python
x_train_new = pd.DataFrame(x_train_sc, columns=x_train.columns)
x_test_new = pd.DataFrame(x_test_sc, columns=x_test.columns)
```

```python
np.round(x_train_new.describe(),2)
```
