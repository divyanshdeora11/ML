```python
# Task Covid dataset convert into categorical to numerical,
# missing data handle
# df --> x,y,x_train, x_test, y_train, y_test
```

```python
covid_df = pd.read_csv('/content/covid_toy.csv')
```

```python
covid_df.head(2)
```

```python
covid_df.info()
```

```python
covid_df.isna().sum()
```

```python
covid_df['fever'] = covid_df['fever'].fillna(covid_df['fever'].mean())
```

```python
covid_df.isna().sum()
```

```python
covid_df['gender'] = covid_df['gender'].map({'Female':1 , 'Male':0})
```

```python
covid_df['cough'] = covid_df['cough'].map({'Mild':0, 'Strong' : 1})
```

```python
covid_df['has_covid'] = covid_df['has_covid'].map({'Yes':1, 'No':0})
```

```python
covid_df.dtypes
```

```python
covid_df = covid_df.drop(columns=['city'])
```

```python
covid_x = covid_df.drop(columns=['has_covid'])
print(covid_x)
```

```python
covid_y = covid_df['has_covid']
print(covid_y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
covid_x_train, covid_x_test, covid_y_train, covid_y_test = train_test_split(covid_x,covid_y,test_size=0.2,random_state=42)
```

```python
print("Total Shape Of covid_df:", covid_df.shape)
print('----*******************----')
print("Total Input Data Shape:", covid_x.shape)
print("x_train Data Shape:", covid_x_train.shape)
print("x_test Data Shape:", covid_x_test.shape)
print('----*******************----')
print("Total Output Data Shape:", covid_y.shape)
print("y_train Data Shape:", covid_y_train.shape)
print("y_test Data Shape:", covid_y_test.shape)
```
