```python
# Encoding --> Transform your categorical data into numerical data
# (1). Label --> 1-1 column encode, we cannot direct select the value of subcategory
# (2). Ordinal --> multi-columns encode, we can select the value of sub_categories
# (3). OneHot -->
# has_covid --> ['Yes', 'No'] --> New Columns --> has_covid_yes, has_covid_no
# column[5 sub-categories] --> 1 sub_category remove --> 4 column -->
# save from the issue of multi-collinearity. It is method of sklearn
# (4). get_dummies --> it is the method of pandas
```
