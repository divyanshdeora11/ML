# Multivarite Analysis

```python
sns.scatterplot(x = 'total_bill', y = 'tip', data=df_tips, hue = df_tips['sex'])
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', data = df_tips,
                hue = df_tips['sex'], style = df_tips['smoker'])
```

  hue -> style -> color -> size  -->

```python
# HeatMap(categorical - categorical)
p = pd.crosstab(df_tips['day'], df_tips['time'])
p
```

```python
sns.heatmap(p)
```

```python
p1 = pd.crosstab(df_tips['sex'], df_tips['smoker'])
p1
```
